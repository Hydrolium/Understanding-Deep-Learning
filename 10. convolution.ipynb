## 합성 곱
- 곱하는 배열을 뒤집은 다음 원본 배열에 계단식으로 곱을 수행하여 더한다.
$$\begin{align*}
x &= [2, 8, 3, 7, 1, 2, 0, 4, 5] \\
w &= [2, 1, 5, 3] \\
w' &= [3, 5, 1, 2]
\end{align*}$$
$$\begin{array}{r|ccccccccc|c}
x & 2 & 8 & 3 & 7 & 1 & 2 & 0 & 4 & 5 & \text{Result} \\
\hline
w' & 3 & 5 & 1 & 2 & & & & & & \rightarrow 63 \\
w' & & 3 & 5 & 1 & 2 & & & & & \rightarrow 48 \\
w' & & & 3 & 5 & 1 & 2 & & & & \rightarrow 49 \\
w' & & & & 3 & 5 & 1 & 2 & & & \rightarrow 28 \\
w' & & & & & 3 & 5 & 1 & 2 & & \rightarrow 21 \\
w' & & & & & & 3 & 5 & 1 & 2 & \rightarrow 20 \\
\end{array}$$
$$\therefore y = [63, 48, 49, 28, 21, 20]$$

```python
# numpy 로 합성곱
import numpy as np

w = np.array([2, 1, 5, 3])
x = np.array([2, 8, 3, 7, 1, 2, 0, 4, 5])

w_r = np.flip(w)

for i in range(6):
    print(np.dot(x[i:i+4], w_r))
```

```python
# scipy 로 합성곱
from scipy.signal import convolve
a = convolve(x, w, mode='valid')
print(a)
```

## 교차 상관
- 합성곱 신경망(CNN)은 합성곱이 아니라 교차 상관을 사용함.
- 교차 상관은 곱하는 배열을 뒤집지 않는다.
$$\begin{align*}
x &= [2, 8, 3, 7, 1, 2, 0, 4, 5] \\
w &= [2, 1, 5, 3]
\end{align*}$$
$$\begin{array}{r|ccccccccc|c}
x & 2 & 8 & 3 & 7 & 1 & 2 & 0 & 4 & 5 & \text{Result} \\
\hline
w & 2 & 1 & 5 & 3 & & & & & & \rightarrow 48 \\
w & & 2 & 1 & 5 & 3 & & & & & \rightarrow 57 \\
w & & & 2 & 1 & 5 & 3 & & & & \rightarrow 24 \\
w & & & & 2 & 1 & 5 & 3 & & & \rightarrow 25 \\
w & & & & & 2 & 1 & 5 & 3 & & \rightarrow 16 \\
w & & & & & & 2 & 1 & 5 & 3 & \rightarrow 39 \\
\end{array}$$
$$\therefore y = [48, 57, 24, 25, 16, 39]$$

- 곱하는 배열(가중치)이 처음부터 무작위로 초기화되어있기 때문에 뒤집는 과정이 필요하지 않음.


## 패딩, 스트라이드
- 원본 배열이 합성곱 연산에 참여하는 정도를 조절하는 방법
- 위에서 $x_0 = 2$ 는 연산에 1번만 참여하고, $x_3 = 7$ 는 연산에 4번 참여함.

#### 패딩(padding)
- 입력 배열의 가장자리에 0을 추가
---
##### 풀 패딩(full padding)
- 모든 원본 배열의 원소가 연산에 참여
- 주로 $\color{red}{0}$ 패딩(제로 패딩)을 추가한다
$$\begin{array}{r|ccccccccccccccc|c}
x & \color{red}{0} & \color{red}{0} & \color{red}{0} & 2 & 8 & 3 & 7 & 1 & 2 & 0 & 4 & 5 & \color{red}{0} & \color{red}{0} & \color{red}{0} & \text{Result} \\
\hline
w & \color{gray}{2} & \color{gray}{1} & \color{gray}{5} & 3 & & & & & & & & & & & & \rightarrow 6 \\
w & & \color{gray}{2} & \color{gray}{1} & 5 & 3 & & & & & & & & & & & \rightarrow 34 \\
w & & & \color{gray}{2} & 1 & 5 & 3 & & & & & & & & & & \rightarrow 51 \\
w & & & & 2 & 1 & 5 & 3 & & & & & & & & & \rightarrow 48 \\
w & & & & & 2 & 1 & 5 & 3 & & & & & & & & \rightarrow 57 \\
w & & & & & & 2 & 1 & 5 & 3 & & & & & & & \rightarrow 24 \\
w & & & & & & & 2 & 1 & 5 & 3 & & & & & & \rightarrow 25 \\
w & & & & & & & & 2 & 1 & 5 & 3 & & & & & \rightarrow 16 \\
w & & & & & & & & & 2 & 1 & 5 & 3 & & & & \rightarrow 39 \\
w & & & & & & & & & & 2 & 1 & 5 & \color{gray}{3} & & & \rightarrow 29 \\
w & & & & & & & & & & & 2 & 1 & \color{gray}{5} & \color{gray}{3} & & \rightarrow 13 \\
w & & & & & & & & & & & & 2 & \color{gray}{1} & \color{gray}{5} & \color{gray}{3} & \rightarrow 10 \\

\end{array}$$
##### 세임 패딩(same padding)
- 출력 배열의 길이가 원본 배열의 길이와 같아지도록 패딩 추가
$$\begin{array}{r|cccccccccccc|c}
x & \color{red}{0} & \color{red}{0} & 2 & 8 & 3 & 7 & 1 & 2 & 0 & 4 & 5 & \color{red}{0} & \text{Result} \\
\hline
w & \color{gray}{2} & \color{gray}{1} & 5 & 3 & & & & & & & & & \rightarrow 34 \\
w & & \color{gray}{2} & 1 & 5 & 3 & & & & & & & & \rightarrow 51 \\
w & & & 2 & 1 & 5 & 3 & & & & & & & \rightarrow 48 \\
w & & & & 2 & 1 & 5 & 3 & & & & & & \rightarrow 57 \\
w & & & & & 2 & 1 & 5 & 3 & & & & & \rightarrow 24 \\
w & & & & & & 2 & 1 & 5 & 3 & & & & \rightarrow 25 \\
w & & & & & & & 2 & 1 & 5 & 3 & & & \rightarrow 16 \\
w & & & & & & & & 2 & 1 & 5 & 3 & & \rightarrow 39 \\
w & & & & & & & & & 2 & 1 & 5 & \color{gray}{3} & \rightarrow 29 \\
\end{array}$$
$$\begin{array}{ccccccccccccc}
