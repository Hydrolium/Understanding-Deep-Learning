## 손실함수
- 실제값과 예측값의 차이를 수치화해주는 함수
- 오차가 클수록 손실 함수의 값이 크다.
1) **MSE**(Mean Squared Error)
    - 회귀문제에서 사용
2) **이진 크로스 엔트로피**(Binary Cross Entropy)
    - 이진 분류 문제에서 사용
    - 출력층에서 시그모이드 함수를 사용하는 이진 분류에서 사용
3) **카테고리별 크로스 엔트로피**(Categorical Cross Entropy)
    - 범주형 교차 엔트로피라고도 부름.
    - 다중 클래스 분류 문제에서 사용
    - 출력층에서 소프트맥스 함수를 사용하는 다중 클래스 분류에서 사용
      

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential

model = Sequential()

# MSE 사용
model.compile(optimizer='adam', loss='mse', metrics=['mse'])
model.compile(optimizer='adam', loss=tf.losses.MeanAbsoluteError(), metrics=['mse'])

# 이진 크로스 엔트로피
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
model.compile(optimizer='adam', loss=tf.keras.losses.BinaryCrossentropy(), metrics=['acc'])

# 카테고리별 크로스 엔트로피
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['acc'])
model.compile(optimizer='adam', loss=tf.keras.losses.CategoricalCrossentropy(), metrics=['acc'])

# 카테고리별 크로스 엔트로피 - 원-핫 인코딩 과정을 생략하고 정수값을 가진 레이블에 대한 다중 클래스 분류 수행
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['acc'])
model.compile(optimizer='adam', loss=tf.keras.losses.SparseCategoricalCrossentropy(), metrics=['acc'])

# 그 외 손실 함수: https://www.tensorflow.org/api_docs/python/tf/keras/losses
```

## 경사 하강법
- 배치는 가중치 등의 매개변수 조정에 사용하는 데이터의 양
- **에포크**: 전체 데이터에 대한 한 번의 훈련 횟수
1) **배치 경사 하강법**(Batch Gradient Descent)
    - 전체 훈련 데이터셋을 사용하여 매개변수를 업데이트하는 방법
    - 1 에포크에 모든 매개변수 업데이트를 한번 수행
    - 계산 비용이 높고 메모리 사용량이 많음
    - 배치 크기가 전체 데이터셋과 같음
2) **확률적 경사 하강법**(Stochastic Gradient Descent)
    - 훈련 데이터셋에서 무작위로 선택된 하나의 샘플을 사용하여 매개변수를 업데이트하는 방법
    - 1 에포크에 매개변수 업데이트를 여러 번(데이터 수만큼) 수행
    - 계산 비용이 낮고 메모리 사용량이 적음
    - 배치 크기가 1
3) **미니 배치 경사 하강법**(Mini-Batch Gradient Descent)
    - 훈련 데이터셋에서 무작위로 선택된 작은 배치를 사용하여 매개변수를 업데이트하는 방법
    - 1 에포크에 (전체 데이터 // 배치 크기) 번만큼 매개변수 업데이트를 수행
    - 계산 비용과 메모리 사용량이 적절히 균형을 이루는 방법
    - 배치 크기가 2 이상 전체 데이터셋보다 작은 값

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 오류 방지용 이진 분류 아무값이나 입력
X_train = np.array([[1,2,3], [1,2,3]])
y_train = np.array([[0], [1]])

model = Sequential([
    Dense(8, input_dim=3, activation='relu'),
    Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy')

# 배치 경사 하강법
model.fit(X_train, y_train, batch_size=len(X_train))

# 확률적 경사 하강법
model.fit(X_train, y_train, batch_size=1)

# 미니 배치 경사 하강법
model.fit(X_train, y_train, batch_size=128) # 기본값: 32
```

## 옵티마이저(Optimizer)
1) **모멘텀**(Momentum)
    - 관성의 개념을 도입하여, 이전 업데이트의 방향과 크기를 고려하여 매개변수를 업데이트하는 방법
    - 매개변수 업데이트에 가속도를 추가하여, 최적점에 빠르게 수렴하도록 돕는 방법
    - 로컬 미니멈(Local Minimum)에서 벗어나 글로벌 미니멈(Global Minimum)으로 갈 수 있도록 도움.
2) **아다그라드**(Adagrad)
    - 학습률을 매개변수마다 적응적으로 조정하는 방법
    - 자주 업데이트되는 매개변수의 학습률을 감소시키고, 드물게 업데이트되는 매개변수의 학습률을 증가시킴
    - 학습률이 점점 감소하여, 최적점에 도달하기 전에 학습이 멈출 수 있음(학습률이 지나치게 떨어짐)
3) **알엠에스프롭**(RMSProp)
    - Adagrad의 단점을 보완한 방법
    - 학습률이 점점 감소하는 문제를 해결하기 위해, 다른 수식(과거의 기울기 제곱의 지수 이동 평균)을 사용하여 학습률을 조정함
    - 학습률이 일정하게 유지되도록 하여, 최적점에 도달할 때까지 학습이 계속되도록 함
4) **아담**(Adam)
    - 모멘텀과 RMSProp의 아이디어를 결합한 방법

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential

# 모멘텀 사용
tf.keras.optimizers.SGD(lr=0.01, momentum=0.9)

# 아다그라드 사용
tf.keras.optimizers.Adagrad(lr=0.01, epsilon=1e-6)

# 알엠에스프롭 사용
tf.keras.optimizers.RMSprop(lr=0.001, rho=0.9, epsilon=1e-06)

# 아담 사용
# tf.keras.optimizers.Adam(lr=0.001, beta_1=0.9, beta_2=0.999, epsilon=None, decay=0.0, amsgrad=False) # 버전 차이로 사용 불가. 시험용으로 알아만 놓기
tf.keras.optimizers.Adam(learning_rate=0.001, beta_1=0.9, beta_2=0.999, epsilon=None, amsgrad=False)


# 옵테마이저 적용
model = Sequential()

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['acc'])
# 또눈
adam = tf.keras.optimizers.Adam(learning_rate=0.001, beta_1=0.9, beta_2=0.999, epsilon=None, amsgrad=False)
model.compile(optimizer=adam, loss='categorical_crossentropy', metrics=['acc'])
```
