## 행렬곱을 통한 순전파
- 3입력, 2출력 인공 신경망
- param의 갯수 8 = ($w_i$ 6개 + $b_i$ 2개)
$$\begin{bmatrix}x_1 & x_2 & x_3\end{bmatrix} \cdot \begin{bmatrix}w_{11} & w_{12} \\ w_{21} & w_{22} \\ w_{31} & w_{32} \end{bmatrix} + \begin{bmatrix}b_1 & b_2 \end{bmatrix} = \begin{bmatrix}y_1 & y_2 \end{bmatrix}$$
$$\begin{align*}
h_1 &= x_1 w_{11} + x_2 w_{21} + x_3 w_{31} + b_1 \\
h_2 &= x_1 w_{12} + x_2 w_{22} + x_3 w_{32} + b_2 \\
[y_1, y_2] &= \text{softmax}([h_1, h_2]) \\
Y &= XW + B
\end{align*}$$


```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()

# 4입력 8출력
model.add(Dense(8, input_dim=4, activation='relu'))

# 이어서 8출력
model.add(Dense(8, activation='relu'))

# 이어서 3출력
model.add(Dense(3, activation='softmax'))


# 입력4 -> 4은닉8 -> 8은닉3 -> 3출력3
```

$X_{m \times n} \times W_{n \times j} + B_{m \times j} = Y_{m \times j}$  

1. 입력4 -> 4은닉8: $X_{1 \times 4} \times W_{4 \times 8} + B_{1 \times 8} = Y_{1 \times 8}$
2. 4은닉층8 -> 8은닉3: $X_{1 \times 8} \times W_{8 \times 3} + B_{1 \times 3} = Y_{1 \times 3}$
3. 8은닉3 -> 3출력3: $X_{1 \times 3} \times W_{3 \times 3} + B_{1 \times 3} = Y_{1 \times 3}$
